### 26

```python
# Input:
# n -> Integer, satr -> String
#------------------------------
# Output:
# yangi_satr -> String

n = int(input("N ni kiriting: "))
satr = input("Satrni kiriting: ")
yangi_satr = ""

satr_uzunlik = len(satr)

if satr_uzunlik > n:
    oriqcha_uzunlik = satr_uzunlik - n 
    yangi_satr = satr[oriqcha_uzunlik:]
elif satr_uzunlik < n:
    oriqcha_uzunlik = n - satr_uzunlik
    for i in range(oriqcha_uzunlik):
        yangi_satr += "."
    yangi_satr += satr

print(f"Natija: {yangi_satr}")
```
